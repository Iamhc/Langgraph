```python
from langgraph.graph import StateGraph
from typing import TypedDict
```

```python
class bmi(TypedDict):
    weight_:float
    height_:float
    result:float
```

```python
def calculate(state:bmi)->bmi:
    weight:state['weight_']
    height:state['height_']
    state['result']=weight/height*height
```
